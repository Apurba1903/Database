```python
import chardet

# Detect encoding
with open("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-35/movies.csv", 'rb') as f:
    encoding = chardet.detect(f.read())['encoding']
    print(f"Detected encoding: {encoding}")
```

```python
from sqlalchemy import create_engine
import pandas as pd


engine = create_engine(f'mysql+pymysql://root:Apurba1903@127.0.0.1:3306/sql_cx_live')
```

```python
groups = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-34/groups.csv")
membership = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-34/membership.csv")
movies = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-35/movies.csv", sep=None, encoding=encoding, engine='python', on_bad_lines='skip')
person1 = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-34/person1.csv")
person2 = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-34/person2.csv")
students = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-34/students.csv")
titanic = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-22/titanic.csv")
users = pd.read_csv("C:/Users/ACER/Desktop/DSMP1/Datasets/datasets-session-34/users.csv")
```

```python
groups.to_sql('groups', con=engine, index=False, if_exists='append', chunksize=1000)
membership.to_sql('membership', con=engine, index=False, if_exists='append', chunksize=1000)
movies.to_sql('movies', con=engine, index=False, if_exists='append', chunksize=1000)
person1.to_sql('person1', con=engine, index=False, if_exists='append', chunksize=1000)
person2.to_sql('person2', con=engine, index=False, if_exists='append', chunksize=1000)
students.to_sql('students', con=engine, index=False, if_exists='append', chunksize=1000)
titanic.to_sql('titanic', con=engine, index=False, if_exists='append', chunksize=1000)
users.to_sql('users', con=engine, index=False, if_exists='append', chunksize=1000)
```
