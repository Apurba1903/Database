```python
import pandas as pd
from sqlalchemy import create_engine


drug = pd.read_csv("Drug/Drug.csv")
drug_clean = pd.read_csv("Drug/Drug_clean.csv")

connection_string = f"mysql+pymysql://root:{PASSWORD}@localhost:3306/drug"
engine = create_engine(connection_string)

drug.to_sql("drug", con=engine, if_exists='replace', index=False)
drug_clean.to_sql("drug_clean", con=engine, if_exists='replace', index=False)
```
