```python
import pandas as pd
from sqlalchemy import create_engine


nw_employees = pd.read_csv("NorthWind-half/nw_employees.csv", on_bad_lines='skip', engine='python')
nw_order_details = pd.read_csv("NorthWind-half/nw_order_details.csv", on_bad_lines='skip', engine='python')
nw_orders = pd.read_csv("NorthWind-half/nw_orders.csv", on_bad_lines='skip', engine='python')
nw_products = pd.read_csv("NorthWind-half/nw_products.csv", on_bad_lines='skip', engine='python')
nw_suppliers = pd.read_csv("NorthWind-half/nw_suppliers.csv", on_bad_lines='skip', engine='python')

connection_string = f"mysql+pymysql://root:{PASSWORD}@localhost:3306/northwind_half"
engine = create_engine(connection_string)

nw_employees.to_sql("nw_employees", con=engine, if_exists='replace', index=False)
nw_order_details.to_sql("nw_order_details", con=engine, if_exists='replace', index=False)
nw_orders.to_sql("nw_orders", con=engine, if_exists='replace', index=False)
nw_products.to_sql("nw_products", con=engine, if_exists='replace', index=False)
nw_suppliers.to_sql("nw_suppliers", con=engine, if_exists='replace', index=False)
```
